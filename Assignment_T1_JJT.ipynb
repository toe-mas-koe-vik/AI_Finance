#### How API Calls Work

When you chat with an LLM through a web interface, here's what happens behind the scenes:

1. **You type a message** → Your browser sends it to the server
2. **Server receives request** → Authenticates you via API key
3. **LLM processes** → Model generates a response
4. **Response returned** → Sent back to your browser

````python
import os
from openai import OpenAI
from getpass import getpass

# Securely input your API key
os.environ['OPENROUTER_API_KEY'] = getpass('Enter your OpenRouter API key: ')

# Initialize the OpenAI client with OpenRouter endpoint
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ['OPENROUTER_API_KEY'],
)

DEFAULT_MODEL = "google/gemini-3-flash-preview"  # Fast, cost-effective model
````

````python
def call_llm(prompt, model=DEFAULT_MODEL):
    """
    Call the OpenRouter API with the given prompt.

    Args:
        prompt: The text prompt to send to the model
        model: The model identifier (default: google/gemini-3-flash-preview)

    Returns:
        The model's response text, or None if there was an error
    """
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}]
        )
        return response.choices[0].message.content
    except Exception as e:
        print(f"API Error: {e}")
        return None


def call_llm_structured(prompt, schema, model=DEFAULT_MODEL):
    """
    Call the OpenRouter API with JSON schema enforcement.

    This ensures the model returns responses in a specific format,
    which is useful for classification tasks where you need exact category names.

    Args:
        prompt: The text prompt to send to the model
        schema: JSON schema defining the response format
        model: The model identifier

    Returns:
        Parsed JSON response as a dictionary, or None if there was an error
    """
    import json
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            response_format={
                "type": "json_schema",
                "json_schema": schema
            }
        )
        return json.loads(response.choices[0].message.content)
    except Exception as e:
        print(f"API Error: {e}")
        return None
````
